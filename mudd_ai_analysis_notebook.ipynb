```python
# Section: Data Cleaning & Preprocessing
# Reusable helpers for cleaning, inference, and type conversion.


def drop_null_columns(df, threshold=0.9):
    """Drop columns with a high share of missing values."""
    missing_ratio = df.isna().mean()
    return df.drop(columns=missing_ratio[missing_ratio > threshold].index)


def convert_to_datetime(df, columns):
    """Convert selected columns to datetime dtype."""
    for column in columns:
        if column in df.columns:
            df[column] = pd.to_datetime(df[column], errors="coerce")
    return df


def normalize_text(df, columns):
    """Lowercase text columns and trim whitespace."""
    for column in columns:
        if column in df.columns:
            df[column] = df[column].astype(str).str.strip().str.lower()
    return df

print("Data cleaning helpers are defined and ready to use.")
```
